# Resumen del Capítulo 1: *Partes de una bomba — ¿Qué es un modelo?*

En este capítulo, Cathy O’Neil introduce el concepto de **modelos matemáticos** y explica que, aunque en teoría deberían ser herramientas objetivas y justas, en la práctica suelen **reproducir sesgos y desigualdades**. 

Los modelos se construyen para simplificar la realidad: **reciben datos de entrada, aplican reglas y producen salidas**, las cuales afectan decisiones reales sobre personas. Cuando estos modelos son opacos, difíciles de cuestionar y afectan a grandes poblaciones, se convierten en lo que la autora llama **Armas de Destrucción Matemática (ADM)**.

---

## Ejemplo 1: Reforma educativa en Washington D.C.

En 2007, el alcalde Adrian Fenty contrató a Michelle Rhee para mejorar el sistema educativo de la ciudad.  
Rhee implementó un sistema de evaluación de profesores llamado **IMPACT**, que calificaba el desempeño docente con un modelo estadístico. 

- Al final del curso 2009-2010, se despidió al **2 % de los profesores con peor puntuación**, y al año siguiente a otro 5 %.
- El problema: el modelo era opaco, difícil de comprender incluso para los propios docentes y no consideraba factores externos (como pobreza, contexto familiar o alumnos transferidos de otras escuelas).
- Resultado: se crearon incentivos perversos, como **inflar artificialmente las notas de los alumnos** para mejorar el puntaje del profesor y evitar despidos.

Este caso muestra cómo un modelo bien intencionado puede convertirse en un **ADM**: es **opaco**, **injusto** y **a gran escala**, afectando la vida de miles de docentes y estudiantes.

---

## Ejemplo 2: Modelos de reincidencia en el sistema penitenciario

O’Neil menciona los modelos de **predicción de reincidencia**, usados para recomendar sentencias y libertad condicional.

- Funcionan a partir de **cuestionarios** que preguntan sobre entorno familiar, situación laboral y vecindario.
- Aunque parecen objetivos, en realidad **refuerzan desigualdades estructurales**, ya que factores como el código postal correlacionan con pobreza y raza.
- Así, personas de comunidades vulnerables reciben **condenas más largas** solo por su contexto, no por sus actos individuales.

---

## Ideas Clave del Capítulo

- Los modelos matemáticos no son neutrales: **heredan los sesgos de quienes los diseñan**.
- Una **ADM** tiene tres características:
  1. **Opacidad** (nadie fuera del sistema sabe cómo funciona).
  2. **Escala** (afecta a muchas personas).
  3. **Daño** (produce consecuencias negativas, reforzando desigualdad).
- Si no se regulan, estas herramientas pueden convertirse en un **círculo vicioso** que castiga repetidamente a los mismos grupos.

---

> **Conclusión:** El capítulo invita a reflexionar sobre el poder que tienen los algoritmos en decisiones críticas, y cómo es necesario cuestionar su transparencia, justicia y el impacto que tienen en la sociedad.


# Resumen del Capítulo 2: *Conmocionada — Mi viaje hacia el desencanto*

En este capítulo, Cathy O’Neil narra su **historia personal** y cómo pasó de ser una entusiasta de las matemáticas puras a una crítica de los modelos algorítmicos usados en la vida real. Explica su transición de la academia al mundo de las finanzas y, finalmente, al activismo.

---

## De la Academia a Wall Street

- O’Neil era profesora de matemáticas en Barnard College, especializada en **geometría algebraica aritmética**.  
- En 2007, dejó la universidad para trabajar en el fondo de cobertura **D.E. Shaw** como **científica de datos**, donde diseñaba modelos financieros para valorar productos derivados.

Al principio, le fascinaba la idea de que las matemáticas pudieran **predecir el futuro financiero** y optimizar ganancias. Sin embargo, pronto empezó a notar que esos modelos:

- **No representaban la realidad completa**, sino solo lo que se podía medir.
- **Excluían riesgos sistémicos** que después explotaron en la crisis de 2008.
- Eran usados para vender productos financieros complejos que terminaron afectando negativamente a millones de personas.

---

## Desencanto y Salida de las Finanzas

La crisis de 2008 fue el punto de quiebre: O’Neil vio que los modelos financieros que ayudó a construir **contribuyeron al colapso del mercado**.  
Además, observó que:

- Las pérdidas fueron **socializadas** (rescate a bancos), mientras que las ganancias seguían siendo **privadas**.
- Los modelos **beneficiaban a los ya poderosos** y perjudicaban a los más vulnerables, sin que existiera una rendición de cuentas.

Esto la llevó a abandonar las finanzas en 2011. Se unió al movimiento **Occupy Wall Street**, donde empezó a reflexionar sobre el poder de los datos y la necesidad de mayor transparencia en los modelos.

---

## De la Ciencia de Datos al Activismo

Después de su experiencia en Wall Street, O’Neil trabajó en **startups tecnológicas** diseñando algoritmos de recomendación para e-commerce. Allí notó que los mismos problemas persistían:

- Los algoritmos podían **reforzar desigualdades** al mostrar productos distintos según el perfil del usuario.
- La lógica de optimización no buscaba el bienestar del cliente, sino **maximizar clics y compras**.

Esto la motivó a investigar cómo estos modelos afectan la educación, el empleo, la justicia penal y otros ámbitos. De ahí nace el concepto de **Armas de Destrucción Matemática (ADM)**.

---

## Ideas Clave del Capítulo

- **Los modelos no son neutrales:** reflejan los incentivos de quienes los crean.
- La autora pasó de ver las matemáticas como un refugio ordenado a entender que, aplicadas sin ética, pueden amplificar desigualdades.
- Este capítulo es una **introducción personal** al viaje intelectual de O’Neil y prepara el terreno para los ejemplos concretos de ADM que explora en los capítulos siguientes.

---

> **Conclusión:** Cathy O’Neil nos muestra que incluso los datos y modelos que parecen objetivos pueden causar daño cuando se usan sin responsabilidad. La transparencia y la rendición de cuentas son esenciales para evitar que los algoritmos se conviertan en herramientas de opresión.


# Resumen del Capítulo 3: *Carrera armamentística — Ir a la universidad*

En este capítulo, Cathy O’Neil examina cómo las **Armas de Destrucción Matemática (ADM)** afectan la **educación superior** en Estados Unidos, particularmente en el acceso a la universidad y en el sistema de rankings. La autora explica que los modelos que supuestamente ayudan a mejorar la educación en realidad han creado un **sistema desigual** que castiga a los más pobres y premia a los privilegiados.

---

## Ranking de Universidades: Incentivos Perversos

- Los **rankings universitarios**, como el de *U.S. News & World Report*, utilizan fórmulas que miden reputación académica, tasas de graduación, gasto por estudiante, y otros factores.
- Estos rankings se convirtieron en el **principal criterio** para que los estudiantes elijan dónde postular.
- Las universidades, para subir en la clasificación, priorizan **estrategias que mejoran su puntuación** en lugar de enfocarse en el aprendizaje real.

### Ejemplo: Manipulación de Admisiones

- Algunas universidades **rechazan estudiantes demasiado buenos** que probablemente no se matricularán, para mejorar su tasa de aceptación.
- También incrementan sus gastos por alumno artificialmente (por ejemplo, subiendo sueldos de administrativos) porque el ranking lo considera un indicador positivo.

---

## Universidades con Ánimo de Lucro

O’Neil dedica una sección a las universidades privadas con fines de lucro (for-profit colleges), que se **aprovechan de los estudiantes de bajos recursos**.

- Estas instituciones gastan más en marketing que en enseñanza, atrayendo a personas desesperadas por mejorar sus oportunidades laborales.
- Prometen **grandes salarios** a cambio de costosos programas financiados con deuda estudiantil.
- Sin embargo, sus graduados a menudo no consiguen los empleos prometidos y quedan atrapados en **deudas impagables**, sin poder declararse en bancarrota para aliviarlas.

---

## Modelos de Riesgo y Becas

- Los modelos de datos permiten a las universidades **segmentar a los estudiantes** y cobrarles según su disposición a pagar.
- Esto significa que dos estudiantes en el mismo programa pueden pagar precios muy distintos, no por su capacidad académica sino por **predicciones algorítmicas sobre su sensibilidad al precio**.
- De esta forma, los modelos refuerzan la desigualdad: los estudiantes con menos recursos acaban pagando proporcionalmente más.

---

## Ideas Clave del Capítulo

- Los rankings universitarios y los algoritmos de marketing educativo son **ADMs**: son opacos, de gran escala y pueden causar daños duraderos.
- En lugar de ampliar oportunidades, muchos modelos convierten la educación superior en una **carrera armamentística** en la que las universidades compiten por prestigio, no por formar mejores profesionales.
- Los estudiantes más vulnerables quedan atrapados en **espirales de deuda**, mientras las universidades mejor posicionadas aumentan su exclusividad y sus ingresos.

---

> **Conclusión:** La educación superior, que debería ser un motor de movilidad social, se ha convertido en un sistema que reproduce desigualdades. Las ADM que miden y clasifican universidades incentivan comportamientos que perjudican a los estudiantes de bajos recursos y priorizan la apariencia de excelencia sobre la educación real.

# Clase 1: Introducción a la Ciencia de Datos

## ¿Qué es la Ciencia de Datos?
La **Ciencia de Datos** es la ciencia de **extraer información significativa de los datos**.  
Transforma datos en **información** y luego en **conocimiento**.

### Proceso básico
1. Formular una **pregunta cuantitativa** que pueda responderse con datos.  
2. **Recolectar y limpiar** los datos.  
3. **Analizar** los datos.  
4. **Comunicar** los resultados a una audiencia relevante.  

**Ejemplo:**  
Una empresa de transporte quiere saber *“¿Cuál es la hora del día con mayor demanda de viajes?”*.  
- Recolecta los datos de viajes.  
- Limpia los registros con errores.  
- Analiza las tendencias por hora.  
- Comunica el resultado a gerencia para planificar la flota.  

---

## ¿Qué entendemos por Ciencia?
- Estudio sistemático del mundo físico y natural mediante **observación y experimentación**.  
- Busca descubrir **relaciones de causalidad**: *¿Por qué ocurre algo?*.  
- En problemas complejos, requiere un **enfoque interdisciplinario** para generar hipótesis y explicaciones.  

**Ejemplo:**  
En medicina, entender cómo se transmite un virus requiere **biología, estadística, epidemiología y computación**.

---

## El Círculo de la Imagen
Relación entre **Algoritmos, IA, ML, DL, Big Data y Ciencia de Datos**.

### Diferencias clave

- **Algoritmos**  
  Conjunto de instrucciones automáticas para resolver un problema.  
  **Ejemplo:** Una receta de cocina es un “algoritmo” para preparar un plato.  

- **Inteligencia Artificial (IA)**  
  Área amplia que crea programas capaces de **imitar el comportamiento humano** (razonar, aprender, adaptarse).  
  **Ejemplo:** Un chatbot que responde preguntas de forma natural.  

- **Machine Learning (ML)**  
  Subconjunto de la IA.  
  Se enfoca en algoritmos que **aprenden de los datos** y mejoran con la experiencia.  
  **Ejemplo:** Netflix recomienda películas según lo que viste antes.  

- **Deep Learning (DL)**  
  Subconjunto de ML.  
  Usa **redes neuronales profundas** para procesar grandes volúmenes de datos complejos.  
  **Ejemplo:** Reconocimiento facial en fotos de Facebook.  

- **Big Data**  
  Conjunto de datos con 3 características fundamentales (**VVV**):  
  1. Gran **volumen**  
  2. Gran **velocidad** (tiempo real)  
  3. Gran **variedad** (estructurados y no estructurados)  

  Definición: *Es Big Data cuando no puede almacenarse en una sola unidad de cómputo* (David Crawshaw).  
  **Ejemplo:** Todas las transacciones de Amazon en un día.  

- **Ciencia de Datos**  
  Intersección de todos los anteriores.  
  Usa **algoritmos (IA, ML, DL)** aplicados a grandes volúmenes de datos (Big Data) para **extraer conocimiento útil**.  
  **Ejemplo:** Analizar millones de registros médicos para detectar patrones tempranos de cáncer.  

---

## Proceso de la Ciencia de Datos
1. **Adquisición de datos**  
2. **Limpieza y procesamiento**  
3. **Exploración de datos**  
4. **Modelamiento y análisis**  
5. **Evaluación y validación**  
6. **Visualización**  
7. **Toma de decisiones**  

**Ejemplo aplicado:**  
En marketing digital:  
- Se recolectan clics en anuncios (adquisición).  
- Se eliminan duplicados (limpieza).  
- Se analizan tendencias de clientes (exploración).  
- Se construye un modelo para predecir compras (modelamiento).  
- Se valida con datos nuevos (evaluación).  
- Se grafican los resultados (visualización).  
- Se decide qué anuncio mostrar (toma de decisiones).  

---

## Perfil del Científico/a de Datos
Un(a) científico(a) de datos combina:  
- **Estadística y modelación matemática**  
- **Programación / computación**  
- **Visualización y comunicación**  
- **Conocimiento del campo específico**  
- **Trabajo colaborativo** (equipos interdisciplinarios).  

**Ejemplo:**  
En agricultura:  
- Un estadístico modela el rendimiento de la cosecha.  
- Un programador procesa imágenes satelitales.  
- Un agrónomo interpreta el suelo y clima.  

---

## Ética en la Ciencia de Datos
La Ciencia de Datos **no es neutral**: importa cómo, para qué y con qué datos analizamos.  

### Riesgos éticos
- **Sesgos en los datos**: entrenar con datos sesgados puede discriminar.  
- **Impacto en la sociedad**: los algoritmos pueden reforzar desigualdades.  
- **Visualizaciones engañosas**: presentar datos de forma parcial puede manipular conclusiones.  

### Preguntas clave
- ¿Qué sesgos están presentes?  
- ¿A quién afecta mi modelo?  
- ¿Estoy interpretando los resultados con responsabilidad?  
- ¿Qué pasa si mi modelo se usa fuera de contexto?  

**Ejemplo:**  
Un sistema de puntaje crediticio basado en la dirección de residencia puede discriminar a comunidades pobres.

# Resumen Clase 2 – Datos vs Información y Herramientas

## Datos, Información, Conocimiento, Sabiduría (Pirámide DIKW)

- **Datos**: registros crudos.  
  *Ejemplo*: temperatura = 18°C, 20°C, 22°C.  

- **Información**: datos procesados → promedio = 20°C.  

- **Conocimiento**: interpretar → "La semana estuvo templada".  

- **Sabiduría**: actuar → "Llevaré chaqueta ligera".  

**Frase de John Tukey**:  
“La combinación de datos + ganas de tener respuesta no asegura que los datos la contengan”.  

*Ejemplo*: medir altura de estudiantes no sirve para concluir su nivel socioeconómico → aunque quieras responderlo, los datos no lo permiten.  

---

## Reproducibilidad Científica en Ciencia de Datos

- **En química**: repetir el mismo experimento para validar resultados.  
- **En ciencia de datos**: repetir el análisis completo (mismos pasos, código, parámetros) y obtener los mismos resultados.  

*Ejemplo*:  
Tienes datos de ventas de una empresa.  
Usas un script en Python que calcula ventas promedio, hace una regresión y grafica con matplotlib.  
Otro investigador, con tu código + datos, debería poder correrlo y obtener las mismas tablas y gráficos.  

**Importancia de la reproducibilidad**:  
- Permite verificar resultados.  
- Permite que un equipo cambie parámetros y compare.  
- Genera confianza y colaboración.  

**Cómo lograrlo**:  
- Usar notebooks (Jupyter) para dejar código, resultados y explicaciones en un mismo lugar.  
- Publicar datos y código en GitHub.  
- Evitar pasos “manuales” (hacer clic, copiar a mano). Todo debe ser automático.  
**En ciencia de datos**, datos → información → conocimiento → sabiduría, y la reproducibilidad **significa que cualquier persona pueda tomar tu mismo código y datos, repetir el análisis y obtener resultados idénticos, garantizando transparencia y confianza.**
---

## Herramientas computacionales (ejemplos prácticos)

- **NumPy** → cálculos numéricos.  
  *Ejemplo*: sacar la media de las ventas.  

- **Pandas** → manejar tablas de datos.  
  *Ejemplo*: cargar el archivo CSV con ventas por producto.  

- **GeoPandas** → datos espaciales.  
  *Ejemplo*: ventas por región en un mapa.  

- **Matplotlib / Seaborn** → visualizar datos.  
  *Ejemplo*: gráfico de barras de ventas por mes.  

- **SciPy** → métodos más avanzados: optimización, estadística, integración.  
  *Ejemplo*: encontrar el mejor precio de un producto que maximiza ganancias.  

- **Scikit-learn** → machine learning.  
  *Ejemplo*: predecir ventas futuras a partir de datos históricos.  

- **Jupyter Notebooks** → documentar todo el flujo del análisis con código + gráficos.  

- **GitHub** → guardar versiones de tu proyecto y compartir con tu equipo.  


# Resumen Clase 3 – Tipos de Datos y Almacenamiento

## ¿Por qué es importante conocer los tipos de datos?

- **Elegir la herramienta correcta** → no es lo mismo analizar una tabla (estructurada) que un JSON (semi-estructurado) o un audio (no estructurado).
- **Diseñar soluciones eficientes** → algunos datos son rápidos de procesar (tablas), otros requieren pasos extra.
- **Detectar errores y sesgos** → si sé qué tipo de datos manejo, puedo limpiar y validar mejor.
- **Optimizar el flujo ETL**:
  - **E (Extract):** extraer datos de distintas fuentes.  
  - **T (Transform):** limpiar, dar formato, estructurar.  
  - **L (Load):** cargar en un sistema de almacenamiento (ej: base de datos o DW).  
- **Decisiones correctas** → sin clasificar bien los datos, el análisis puede ser inválido.
- **Habilidad clave** → en el mundo laboral esperan que sepas reconocerlos y trabajarlos.

---

## Cuadro comparativo – Tipos de datos

| **Tipo de datos**   | **¿Cómo se ven visualmente?** | **Características** | **Ejemplos** |
|----------------------|-------------------------------|----------------------|--------------|
| **Estructurados**   | Tablas rectangulares, todas las filas con las mismas columnas. Como una planilla Excel. | Muy organizados, con esquema fijo. Fáciles de analizar con SQL o Pandas. | `.csv`, `.xlsx`, bases SQL, DataFrames de Pandas, Parquet (columnar). |
| **Semi-estructurados** | No siempre en filas/columnas. Datos con etiquetas, llaves o jerarquías. Pueden variar entre registros. | Tienen cierta organización, pero no rígida. Requieren preprocesamiento para “armar una tabla”. | JSON, XML, YAML, logs clave-valor, Avro. |
| **No estructurados** | No hay forma tabular clara. Texto libre, imágenes, audios o videos. | Sin esquema fijo. Muy ricos en información, pero difíciles de procesar. Necesitan herramientas especiales (IA, ML). | Fotos, correos, redes sociales, PDFs escaneados, audios de call center. |

---

## Entonces, ¿qué son Data Warehouse (DW) y Data Lake (DL)?

- No son “tipos de datos”, son **formas de almacenar y organizar datos**.  
- Mientras que los datos pueden ser estructurados, semi o no estructurados, un **DW** o un **DL** son **infraestructuras donde los guardas para poder analizarlos**.

### Explicación con peras y manzanas:

- **Base de datos:** como una libreta donde anotas ventas del día (muy detallada, en tiempo real).  
- **Data Warehouse:** como un libro de resúmenes. Cada cierto tiempo, copias lo importante de varias libretas (ventas, inventario, clientes) y lo resumes en tablas para hacer reportes rápidos. → Se queda con **datos estructurados y resumidos**.  
- **Data Lake:** como una gran bodega donde tiras todo lo que recopilas (libretas completas, fotos de productos, grabaciones de llamadas, comentarios en redes). No resumes nada, los guardas **“en bruto”** para analizarlos después.  

---

## Cuadro comparativo – Data Warehouse vs Data Lake

| **Característica**           | **Data Warehouse (DW)** | **Data Lake (DL)** |
|-------------------------------|--------------------------|---------------------|
| **Tipo de datos que guarda** | Principalmente estructurados, listos para consulta. | Todos: estructurados, semi y no estructurados. |
| **Propósito**                | Reportes, análisis tradicional (ej: BI). | Exploración avanzada, Big Data, Machine Learning. |
| **Cómo se ven**              | Tablas resumidas: ventas por mes, clientes por región. | Carpeta gigante con datos en bruto de múltiples fuentes. |
| **Procesamiento**            | Los datos ya están limpios y resumidos (OLAP). | Los datos se guardan tal cual llegan. Se procesan después. |
| **Velocidad de consulta**    | Muy rápida, porque la info ya está preparada. | Más lenta, porque hay que procesar antes de consultar. |
| **Ejemplo**                  | “Total de ventas por producto en 2024”. | “Analizar sentimientos de clientes en redes sociales sobre productos”. |

---

## Pregunta clave:

**¿Un CSV puede ser un Data Warehouse?**  
- No. Un CSV es un **formato de datos estructurados** (como una tabla).  
- Un **Data Warehouse** en cambio es una **infraestructura** que puede contener miles de tablas en CSV, SQL u otros formatos, pero **organizadas, resumidas y optimizadas para análisis**.  

**¿En qué categoría caen DW y DL?**  
- No son categorías de datos, sino **sistemas de almacenamiento** que pueden contener distintos tipos de datos.  


# Resumen Clase 4 – Datos Geoespaciales

## ¿Qué son los datos geoespaciales?
- Son un tipo especial de datos que representan información de la superficie terrestre.  
- Se organizan en **capas (layers)** que describen distintos aspectos de un mismo lugar.  
  Ejemplo: una capa de calles, una de ríos y otra de uso de suelo.  

Existen dos formas principales de representar estos datos: **raster** y **vectoriales**.  

---

## Datos Raster

### Cómo se ven
- Como una **imagen de pixeles (una cuadrícula)**.  
- **Cada celda (pixel) tiene un valor** que puede ser:
  - **Intensidad de color** (ej: foto satelital en RGB).  
  - **Medida física** (ej: temperatura, altura del terreno, humedad).  

👉 Ejemplo:  
Una foto satelital donde **cada pixel representa la temperatura del suelo en esa ubicación**.  
- Si el pixel es rojo: 35°C  
- Si es azul: 15°C  

### Cómo se generan
- Se “**muestra**” la superficie en intervalos regulares formando una **grilla**.  
- Cada celda de la grilla almacena un valor promedio de lo que ocurre en esa área.  

👉 Ejemplos:  
- Imágenes satelitales **Landsat**.  
- Mapas de elevación digital (DEM).  

### Aplicaciones
- Mapas de calor (temperatura, contaminación, humedad).  
- Altura del terreno (modelos digitales de elevación).  
- Agricultura: detectar zonas más verdes/seca con imágenes multiespectrales.  

### Cuándo usar raster
- Cuando quiero representar algo **continuo**, sin bordes exactos.  
- Ejemplo: temperatura, humedad, contaminación → cambian gradualmente en el espacio.  

### Problemas
- Archivos muy pesados si la resolución (tamaño del pixel) es alta.  
- Se pierde detalle si la resolución es muy baja.  

---

## Datos Vectoriales

### Cómo se ven
Representan objetos **geométricos**:
- **Puntos** → lugares exactos (coordenadas).  
  Ejemplo: ubicación de una farmacia.  
- **Líneas** → conexiones.  
  Ejemplo: carreteras, ríos.  
- **Polígonos** → áreas cerradas.  
  Ejemplo: límites de una comuna, un lago, un parque nacional.  

### Cómo se almacenan
- Cada objeto tiene **atributos asociados**.  
  Ejemplo: un polígono que representa una comuna puede tener:  
  - `nombre = "Santiago"`  
  - `población = 7 millones`  

### Formatos comunes
- `.shp` (Shapefile)  
- `.geojson`  
- `.kml` (Google Earth)  

### Aplicaciones
- Urbanismo: trazar límites de barrios o calles.  
- Política: dibujar límites de comunas y provincias.  
- Transporte: mapas de metro, buses, carreteras.  

### Cuándo usar vectoriales
- Cuando necesito **precisión en objetos con bordes definidos**.  
- Ejemplo: saber dónde empieza y termina un lago, o qué calles forman parte de una comuna.  

### Problemas
- Se vuelven complicados para fenómenos **continuos** (ej: temperatura, contaminación).  
- Necesitan cuidado con el sistema de coordenadas (CRS).  

---

## Raster vs Vectorial – ¿Cuándo usar cada uno?

- **Raster** → cuando quiero ver cómo varía una magnitud en toda la superficie.  
  Ejemplo: mapa de calor de contaminación en Santiago.  

- **Vectorial** → cuando quiero representar objetos definidos con atributos.  
  Ejemplo: polígonos con los límites de cada comuna.  

👉 Lo más común es **combinarlos**:  
- Raster: mapa de contaminación.  
- Vectorial: límites comunales.  
- Resultado: calcular la contaminación promedio en cada comuna.  

---

## Problemas comunes de los datos geoespaciales
- **Sistema de coordenadas (CRS)**: distintos datasets pueden usar proyecciones diferentes.  
  Ejemplo: uno en latitud/longitud y otro en UTM → hay que unificarlos.  
- **Resolución en raster**: un pixel puede representar 1 km² o 10 m²; la elección cambia la precisión.  
- **Archivos pesados**: raster de alta resolución o vectoriales con millones de geometrías.  
- **Falta de consistencia**: distintas fuentes pueden definir distinto un mismo límite (ej: borde de una comuna).  

---

## Ejemplo Integrador

Imagina que eres un planificador urbano:

1. Usas un **raster de temperatura** de la ciudad para ver las “islas de calor”.  
2. Usas **vectoriales de polígonos de las comunas** para calcular en qué comunas hay más calor acumulado.  
3. Combinas ambos para **proponer dónde plantar más áreas verdes**.  

---

## Aclaración importante sobre los **raster**
Cuando decimos que *"cada celda tiene un valor, como la intensidad de color o una medida física"*, se refiere a que:  
- Un **pixel** no es solo una mancha de color, sino que **guarda un dato numérico**.  
- Ese dato puede luego traducirse a color para que lo entendamos visualmente.  

👉 Ejemplos:  
- Imagen satelital de infrarrojo: cada pixel guarda un número que indica la **temperatura del suelo**.  
- Mapa de elevación: cada pixel guarda un número que representa la **altura en metros** de ese punto.  

En resumen: **el color que ves es solo una representación visual del valor numérico real que guarda el pixel**.

# Resumen Clase 7 - Fuentes de datos: API
## ¿Qué es una API?

Una **API (Application Programming Interface)** es como un **mesero en un restaurante**:

- Tú (usuario) no vas a la cocina (la base de datos de eBird).
- Tú le pides al mesero (la API) lo que quieres: "tráeme una lista de avistamientos de pájaros en Chile".
- El mesero lleva tu pedido a la cocina y luego te trae la comida servida (los datos, generalmente en formato **JSON**, que las máquinas entienden fácilmente).

---

## 1. Obtener una key o credenciales de acceso

- Algunas APIs son públicas y cualquiera puede usarlas.
- Pero muchas requieren que te registres y te den una **llave (API key)**.

Piensa en la **API key como tu carnet de socio**:

- Sirve para que el servidor sepa quién eres.
- También pone límites: por ejemplo, "Maca puede hacer 1000 consultas al día".

En **eBird** necesitas crear una cuenta y obtener una **API key**.  
Esa key luego se manda en el **header** (te lo explico más abajo).

---

## 2. Entender cómo hacer consultas a la API (query string)

Cada API tiene **documentación** que te dice cómo pedirle cosas.  
Por ejemplo:

- Quiero avistamientos recientes en Chile.
- Quiero avistamientos de una especie específica (ej: pingüino de Humboldt).

La documentación te dice qué parámetros puedes usar en la consulta:

- `regionCode=CL` (país Chile)
- `speciesCode=humpen` (código de especie)
- `maxResults=50` (cuántos registros quiero)

Ese conjunto de parámetros se llama **query string**.  
Es como el pedido que le das al mesero: "quiero pizza, pero sin queso, y con champiñones".

---

## 3. Enviar solicitud (requests.get())

Cuando le pedimos datos a la API, necesitamos tres piezas principales:

### url
Es la **dirección del recurso**.  
Ejemplo en eBird:  



```python
url = "https://api.ebird.org/v2/data/obs/CL/recent"
```

Eso significa: “dame los avistamientos recientes en Chile”.

### params
Un **diccionario** donde ponemos los parámetros que personalizan la consulta.
Ejemplo:

```python
params = {
   "maxResults": 10
}
```

Esto significa: "dame solo 10 resultados".

Piensa en esto como **personalizar tu pedido.**

### headers
Son las **cabeceras HTTP**: información extra que le pasas a la API.
En eBird, aquí es donde pones tu **API key** para que te reconozca.
Ejemplo:

```python
headers = {
   "X-eBirdApiToken": "tu_api_key_aqui"
}
```

Es como mostrar tu carnet de socio antes de que el mesero te atienda.

### 4. Resultado: el objeto response

Cuando mandas la solicitud, la API responde con un **objeto response.**

Ese objeto tiene los datos en JSON.

Para leer solo los datos, usas .json().

Ejemplo de lo que te puede devolver (muy resumido):

```python
[
  {"speciesCode": "houspa", "comName": "Gorrión común", "obsDt": "2023-05-01", "lat": -33.45, "lng": -70.66},
  {"speciesCode": "rufhum", "comName": "Picaflor rubí", "obsDt": "2023-05-01", "lat": -33.47, "lng": -70.65}
]
```

Eso significa:

Se vio un gorrión común en Santiago.

Se vio un picaflor rubí en otro lugar cercano.

### 5. Transformar en DataFrame

Como JSON es un formato de texto, para trabajar más cómodo en Python lo pasamos a un **DataFrame de pandas** (tablas).
Así puedes analizar, filtrar, graficar.

# Clase 8 - Fuentes de datos: Webscraping

## 1. ¿Qué es Web Scraping?

El **Web Scraping** es una técnica para extraer información de sitios web de forma automática y luego guardarla en un formato estructurado (ej: CSV, JSON, tabla).

- **Con URL directo**: te bajas un archivo ya preparado (CSV, Excel, JSON).
- **Con API**: le pides a un servidor los datos en formato limpio y actualizado.
- **Con Scraping**: la página no te entrega un archivo ni API, entonces tú entras como un “navegador robot” y vas leyendo el código de la página (HTML) para sacar los datos que te interesan.

**Ejemplo**:  
Imagina que una página muestra una tabla con precios de frutas, pero no tiene ni botón de descarga ni API.  
Con Web Scraping programas un “robot” que entra, lee el HTML, encuentra la tabla y se guarda esos precios en un archivo para ti.

---

## 2. ¿Es legal?

En general sí, pero depende del sitio.

- El scraping aumenta el tráfico de la página, y si muchos lo hacen, puede botar el servidor.
- Algunos sitios dicen explícitamente si permiten scraping en un archivo llamado **robots.txt**.

**Ejemplo**: entra a [https://www.wikipedia.com/robots.txt](https://www.wikipedia.com/robots.txt) y verás qué partes permiten o prohíben.

---

## 3. El proceso de Web Scraping

El scraping sigue pasos simples pero ordenados:

1. **Inspeccionar la página web**  
   - Mirar su código HTML (click derecho → “Inspeccionar”).  
   - Ejemplo: ubicar dónde está la tabla de precios en el código.

2. **Acceder a la URL y descargar el HTML**  
   - Usamos `requests.get(url)` para bajar el código HTML completo (el mismo que verías en “Ver código fuente”).

3. **Formatear el HTML (parsear)**  
   - El HTML viene como texto con muchas etiquetas `< >`.  
   - Con la librería **BeautifulSoup** lo convertimos en un árbol fácil de recorrer.

4. **Extraer información útil**  
   - Buscamos en el árbol HTML los elementos que nos interesan (ej: todas las filas de una tabla).  
   - Guardamos esa info en un formato estructurado (ej: lista, DataFrame, CSV).

5. **Repetir si hay varias páginas**  
   - Si los datos están en más de una página (ej: página 1, 2, 3…), hacemos un bucle y repetimos el proceso.

---

## 4. Entendiendo HTML (con peras y manzanas)

El **HTML** es el lenguaje que usan las páginas web.  
Funciona como un **árbol de cajas anidadas**:

- **Etiqueta de inicio**: `<p>`  
- **Etiqueta de fin**: `</p>`  
- **Contenido entre ellas**: texto o más etiquetas.

```python
#<html>
  #<body>
    #<h1>Título de la página</h1>
    #<p>Este es un párrafo con <b>negrita</b>.</p>
    #<table>
      #<tr><td>Manzana</td><td>100</td></tr>
     # <tr><td>Pera</td><td>80</td></tr>
    #</table>
  #</body>
#</html>
```

Lo dejé comentado para que no me salga error.

Con **BeautifulSoup**, podemos decir:

`["Manzana", 100]`, `["Pera", 80]`.

# Clase 10 - Transformación de datos
## 3. Transformación de datos
La transformación de datos es el proceso de **convertir datos crudos en datos adecuados** para el análisis.  
Se realiza porque los datos originales pueden contener errores, estar en formatos incompatibles o requerir simplificación.  
Este paso es clave porque es donde se **genera más valor** dentro del ciclo de ciencia de datos.  

### Pasos principales del proceso
1. **Descubrimiento**: entender los datos disponibles y su estado (profiling).  
2. **Mapeo**: planificar cómo se transformarán los datos.  
3. **Generación del proceso**: escribir la “receta” de transformación (en código o procedimientos).  
4. **Ejecución**: aplicar la transformación.  
5. **Revisión**: verificar la calidad de la información obtenida.  

---

## 4. Data Wrangling
El *data wrangling* es la limpieza, transformación y organización de datos crudos para que queden listos para el análisis.  

### Tareas típicas
- Eliminar nulos y duplicados.  
- Cambiar formatos y tipos de datos.  
- Unir diferentes fuentes.  
- Normalizar valores para hacerlos comparables.  
- Filtrar información relevante.  
- Enriquecer datos agregando información adicional.  

### Diferencias con ETL
- **Data Wrangling**: más flexible, manual o semi-automatizado, enfocado en análisis puntuales.  
- **ETL**: estructurado y automatizado, orientado a mover grandes volúmenes de datos hacia sistemas centralizados.  

### ETL vs ELT
- **ETL**: transforma antes de cargar.  
- **ELT**: carga los datos crudos y luego transforma, más común en *data lakes*.  

---

## 5. Tipos de datos
Cada variable debe tener un **tipo de dato adecuado** para evitar errores en el análisis.  
Ejemplos:  
- Texto → `str`, `object`  
- Enteros → `int`, `int64`  
- Decimales → `float`, `float64`  
- Binario → `bool`  
- Fecha → `datetime`  
- Categorías → `category`  

En la práctica, es necesario revisar y ajustar los tipos de datos al cargar un dataset, ya que muchas veces son mal interpretados automáticamente.  

---

## 6. Manejo de problemas en datos
### Datos faltantes
Se pueden eliminar, imputar con valores constantes, media, mediana, interpolación o mediante algoritmos.  

### Datos duplicados
- Eliminarlos si son registros repetidos completos.  
- Tomar decisiones caso a caso cuando son duplicados parciales (ej. promedio, máximo o mínimo).  

### Datos categóricos
- Representan un conjunto acotado de valores.  
- Necesitan codificación numérica para modelos de ML (ej. *one-hot encoding*).  
- Pueden requerir eliminar, remapear o inferir categorías para corregir inconsistencias.  

### Variables indicativas (dummies)
Transformar categorías en variables binarias (1 o 0) para usarlas en algoritmos.  

### Otras técnicas de limpieza
- Validar consistencia entre campos (ej: edad vs fecha de nacimiento).  
- Revisar uniformidad en formatos y unidades.

## 3. Limpieza de datos
La limpieza de datos consiste en **identificar y corregir** información que sea incorrecta, incompleta, imprecisa, irrelevante o faltante.  
El objetivo es **modificar, reemplazar o eliminar** estos datos para garantizar un análisis confiable.  

Este paso es esencial porque si los errores no se corrigen, se **propagan a lo largo del flujo de trabajo**, afectando la extracción, la exploración, la generación de conocimiento y los reportes finales.  

---

## 4. ¿Dónde limpiar datos?
El proceso de limpieza ocurre entre la **recolección del dataset crudo** y la construcción del **dataset limpio**.  

El dataset limpio es la base para tareas clave como:  
- análisis exploratorio,  
- modelamiento estadístico,  
- algoritmos de machine learning,  
- visualización y comunicación de resultados,  
- toma de decisiones.

## 5. Problemas comunes en los datos
Al trabajar con datasets suelen aparecer distintos problemas que requieren limpieza:  
- **Datos faltantes**, cuando no se dispone de información en algunas observaciones.  
- **Duplicados**, que repiten registros de manera completa o parcial.  
- **Transformaciones de valores**, cuando los datos requieren ajustes de formato o tipo.  
- **Valores nulos**, que representan ausencia de información.  
- **Datos innecesarios**, que no aportan al análisis.  
- **Datos inconsistentes**, que entran en conflicto con la lógica o con otras variables.

## 6. Operaciones comunes en Pandas
Al preparar datos es frecuente tener que **eliminar o renombrar columnas** para mejorar la calidad del dataset.  
Algunas operaciones típicas son:  
- Revisar el listado de columnas,  
- Eliminar aquellas que no son útiles,  
- Seleccionar solo las variables necesarias,  
- Renombrar columnas para mejorar su legibilidad y manipulación posterior.
